**Numpy arrays vs Python Sequences**

*  Numpy arrays have a fixed size at creation , unlike python lists which grow dynamically.

*  Changing the size of an ndim array will create a new array and delete the original.

*  The elements in a Numpy Array are required to be of the same data type and thus will be of the same size in memory.

*  Numpy Arrays facilitates advanced mathematical and other types of operations on large number of data.Typically , such operations are executed more efficiently with less code than is possible using python's built-in sequences.

```python
# Import NumPy library
import numpy as np


# Create a normal Python list
list1 = []

# Loop from 0 to 14 (15 elements total)
for i in range(15):
  list1.append(i)   # Add each number to the list

# Print the Python list
print(list1)


# NumPy arange(start, stop)
# Creates an array from 3 to 14 (15 is excluded)
print(np.arange(3,15))   # Continuous array


# NumPy arange(start, stop, step)
# Starts from 3, ends before 15, step = 2
# Skips numbers → output: 3,5,7,9,11,13
print(np.arange(3,15,2))


# Create NumPy array from 2 to 15
arange_data = np.arange(2,16)   # total elements = 14


# reshape(rows, columns)
# ⚠️ reshape does NOT change data, only changes structure
# Condition: rows × columns = total elements
# Here: 2 × 7 = 14 ✔
arange_data.reshape(2,7)


# Create array from 2 to 16 (step = 2)
# Output: [2,4,6,8,10,12,14,16] → total 8 elements
# reshape into 2 rows and 4 columns → 2 × 4 = 8 ✔
print(np.arange(2,17,2).reshape(2,4))
```

**MATHEMATICAL OPERATIONS IN NUMPY**



*   Scalar Operations --> A scalar value operation will be done with the array or a numeric value(E.g. 1,4)

*   Vector Operations --> The operations will be in between the given number of matrices.




```python
#Mathematical Operations
import numpy as np
array = np.random.random((3,3))*100 #Scalar Operation
array = np.round(array)
array = array.astype('int32')
print(array)

#Scalar Operations with our matrix or Array

print(array + 2)
print(array - 2)
print(array * 2)
print(array / 2)
```

**We can apply scalar operations with the array in terms of arithmetic operations.**

```python
import numpy as np
a = 5
b = 6
print(a > b)
print(a < b)

array = np.random.random((3,3))*100
print(array > 60) #This operation will compare each value with 60 and give true or false respectively.
```

```python
#Vector Operations
import numpy as np
arr_1 = np.arange(1,17).reshape(4,4)
arr_2 = np.arange(17,33).reshape(4,4)
print(arr_1)
print(arr_2)

print(arr_1 + arr_2)
print(arr_2 - arr_1)
print(arr_1 * arr_2)
print(arr_1 / arr_2)

```

**Important functions in Numpy for machine learning and data science**



*   max() --> It is going to be used to calculate the max value inside a matrix or an array.SYNTAX : max() --> np.max(array_name)

*   axis() --> It provides us with the idea about the any operation value inside a specific row or column or we can get a specific value row wise or column wise.

1.   axis = 0 --> Column Wise Operation  
2.   axis = 1 --> Row Wise Operation  


*   min() --> It is going to be used to calculate the min value inside a matrix or an array.SYNTAX : min() --> np.min(array_name)

*   sum() --> Calculate the sum of all of the values of a given matrix.SYNTAX : np.sum(array_name)

*   prod() --> It helps with calculating the product of all values inside the matrix.SYNTAX : np.prod(array_name)



```python
#max()
import numpy as np
arr_1 = np.arange(1,17).reshape(4,4)
print(arr_1)
print(np.max(arr_1))
print(np.max(arr_1,axis = 0))

array = np.round(np.random.random((4,4))*100).astype('int32')
print(array)
print(np.max(array,axis = 0)) #Column Wise Operation
print(np.max(array,axis = 1)) #Row Wise Operation

#min()
print(np.min(arr_1))
print(np.min(array,axis = 0)) #Column Wise Operation
print(np.min(array,axis = 1)) #Row Wise Operation

#sum()
print(np.sum(array))
print(np.sum(array,axis = 0)) #Column Wise Operation
print(np.sum(array,axis = 1)) #Row Wise Operation

#prod()
print(np.prod(array))
print(np.prod(array,axis = 0)) #Column Wise Operation
print(np.prod(array,axis = 1)) #Row Wise Operation


```

**Stacking and Splitting in Array --> Size of the arrays must be the same.**

**Stacking --> Joining arrays together.**


*   Horizontal Stacking --> Row by Row joining is done. --> hstack()
*   Vertical Stacking --> Column by Column joining is done. --> vstack()




```python
#Stacking
import numpy as np
array_1 = np.arange(1,13).reshape(3,4)
array_2 = np.arange(13,25).reshape(3,4)

print(array_1)
print(array_2)

#Horizontal Stacking
horizontal_array = np.hstack((array_1,array_2))
print(horizontal_array)

#Vertical Stacking
vertical_array = np.vstack((array_1,array_2))
print(vertical_array)
```

**Splitting --> Dividing an array into multiple sub-arrays.**

  

*   Horizontal Split --> np.hsplit(array,number of parts you want your array to be splitted at.)

*   Vertical Split --> np.vsplit(array,number of parts you want your array to be splitted at.)



```python
#Splitting
import numpy as np
array_1 = np.arange(1,13).reshape(3,4)

print(array_1)

#Horizontal Split
horizontal_split = np.hsplit(array_1,2)
print(horizontal_split)

#Vertical Split
vertical_split = np.vsplit(array_1,3) # Changed 2 to 3, as the array has 3 rows, to allow for an even split.
print(vertical_split)
```
