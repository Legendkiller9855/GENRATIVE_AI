```python
# TASK - 1 : Basic Class & Object Creation

class Product:
  def __init__(self, name, price, category):
    self.name = name
    self.price = price
    self.category = category

  def get_info(self):
    print(f"The name of the product is : {self.name} and the price is : {self.price} and the category of the given product is : {self.category}")

  def method_apply(self, discount):
    self.price = self.price - (self.price * discount / 100)
    print(f"New price after {discount}% discount is : {self.price}")

dustbin = Product("dustbin", 34, "Home cleanliness")
dustbin.get_info()

dustbin.method_apply(10)

cooler = Product("cooler", 134, "Home Conditioner")
cooler.get_info()

cooler.method_apply(20)
```

```python
#TASK - 2 : Constructor & Encapsulation

class Product:
  def __init__(self, name, price, category):
    self.name = name
    self.__price = price #private attribute
    self.category = category

  #getter method
  def get_price(self):
    print(f"The price of the product {self.name} is : {self.__price}$")

  #setter method
  def set_price(self,new_price):
    if new_price > 0:
     self.__price = new_price
     print(f"The updated price of the {self.name} is {self.__price}$")
    else:
      print("The updated price must be greater than zero.")

cooler = Product("cooler", 134, "Home Conditioner")
cooler.get_price()

cooler.set_price(234)
```

```python
#TASK - 3 : Inheritance(Single-Level)


class Product:
  def __init__(self, name, price, category):
    self.name = name
    self.price = price
    self.category = category

  def get_info(self):
    print(f"The name of the product is : {self.name} and the price is : {self.price} and the category of the given product is : {self.category}")

class ElectronicProduct(Product):
  def __init__(self,name,price,category,warranty_years):
    super().__init__(name,price,category)
    self.warranty_years = warranty_years

  def get_info(self):
    print(f"The name of the product is : {self.name} and the price is : {self.price} and the category of the given product is : {self.category} and the warranty of the product is for : {self.warranty_years} years.")

cooler = ElectronicProduct("cooler",134,"house Conditioner",7)
cooler.get_info()
```

```python
#TASK - 4 : Polymorphism

class Laptop:
  def __init__(self,product):
    self.product = product

  def get_info(self):
    print(f"The name of the product is : {self.product} and is used to play heavy games like GTA and do coding.")


class Mobile:
  def __init__(self,product):
    self.product = product


  def get_info(self):
    print(f"The name of the product is : {self.product} and is used to do normal day-to-day activities.")

mobile = Mobile("mobile")
laptop = Laptop("laptop")

devices = [mobile,laptop]
for device in devices:
  device.get_info()
```

```python
#TASK - 5 : Abstraction(Using Abstract Base Class)

from abc import ABC, abstractmethod

class Payment(ABC):

  @abstractmethod
  def process_payment(self, amount):
    pass


class CreditCardPayment(Payment):

  def process_payment(self, amount):
    print(f"Processing credit card payment of ₹{amount}")

class UPIPayment(Payment):

  def process_payment(self, amount):
    print(f"Processing UPI payment of ₹{amount}")

credit = CreditCardPayment()
upi = UPIPayment()

credit.process_payment(5000)
upi.process_payment(1500)
```

```python
#TASK - 6 : Magic Methods & Operator Overloading

class Product:

  def __init__(self, name, price, category):
    self.name = name
    self.price = price
    self.category = category

  def __str__(self):
    return f"Product({self.name}, {self.price}, {self.category})"

  def __add__(self, other):
    return self.price + other.price

product1 = Product("Laptop", 50000, "Electronics")
product2 = Product("Mobile", 20000, "Electronics")

print(product1)
print(product2)

total_price = product1 + product2
print(f"Total combined price: {total_price}")
```

```python
#TASK - 7 : Mini Project : Simple Inventory System(OOP only)

class Product:

    def __init__(self, name, price, category):
        self.name = name
        self.price = price
        self.category = category

    def __str__(self):
        return f"{self.name} - ₹{self.price} ({self.category})"

    def __add__(self, other):
        return self.price + other.price

class Inventory:

    def __init__(self):
        self.products = []

    def add_product(self, product):
        self.products.append(product)
        print(f"{product.name} added to inventory.")

    def remove_product(self, name):
        for product in self.products:
            if product.name == name:
                self.products.remove(product)
                print(f"{name} removed from inventory.")
                return
        print("Product not found.")

    def get_total_value(self):
        total = 0
        for product in self.products:
            total += product.price
        return total


    def show_all_products(self):
        print("\nAll Products:")
        for product in self.products:
            print(product)


class Store:

    def __init__(self, store_name):
        self.store_name = store_name
        self.inventory = Inventory()

    def add_new_product(self, name, price, category):
        product = Product(name, price, category)
        self.inventory.add_product(product)


    def show_summary(self):
        print(f"\nStore Name: {self.store_name}")
        total_items = len(self.inventory.products)
        total_value = self.inventory.get_total_value()

        print(f"Total Items: {total_items}")
        print(f"Total Value: ₹{total_value}")

        self.inventory.show_all_products()


store = Store("Krishan Mart")

store.add_new_product("Laptop", 50000, "Electronics")
store.add_new_product("Mobile", 20000, "Electronics")
store.add_new_product("Chair", 3000, "Furniture")

store.show_summary()

p1 = Product("Tablet", 15000, "Electronics")
p2 = Product("Headphones", 5000, "Electronics")

print("\nCombined Price using + :", p1 + p2)
```
