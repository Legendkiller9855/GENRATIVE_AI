**In Object Oriented Programming , we are trying to match up with the real-world scenario by using programming.**

**In OOPS , we are going to learn about classes and objects.**

**By a class we mean to provide a blueprint that must remain the same for each and every single object for us.**

**These objects can have different functionalities from each other and different characteristicts from each other as well but they must follow the similar blueprint.(CLASS)**

**By a blueprint or a class what we mean to have a means of providing the functionalities and the characteristicts to our objects.**

```python
# Define a class named Avenger
class Avenger:

  # Constructor method (__init__) runs automatically when an object is created
  # It initializes the attributes of the object
  def __init__(self, name, power, weapon, nationality="American"):

    # self.name stores the name of the Avenger
    self.name = name

    # self.power stores the special power of the Avenger
    self.power = power

    # self.weapon stores the weapon used by the Avenger
    self.weapon = weapon

    # self.nationality stores the nationality of the Avenger
    # "American" is the default value if no nationality is provided
    self.nationality = nationality


  # Method to introduce the Avenger
  # This method prints the details of the Avenger
  def introduction(self):

    # Print the Avenger's name
    print(f"Hello my name is : {self.name}")

    # Print the Avenger's weapon
    print(f"Hello my weapon is : {self.weapon}")

    # Print the Avenger's power
    print(f"Hello my power is : {self.power}")

    # Print the Avenger's nationality
    print(f"Hello my nationality is : {self.nationality}")


# Creating an object of the Avenger class
# Iron_man is an instance (object) of the Avenger class
Iron_man = Avenger("Ironman", "Genius Intellect", "Repulsor Beams")

# Calling the introduction method using the object
Iron_man.introduction()

print(Iron_man.name)

# Key Concepts in This Code

# Class → Blueprint for creating objects

# Constructor (__init__) → Initializes object attributes

# self → Refers to the current object

# Attributes → Variables inside a class (name, power, etc.)

# Method → Function inside a class (introduction)

# Object → Instance of a class (Iron_man)


Captian_America = Avenger("Captian America","Super Strength","Shield","Portugese")
Captian_America.introduction()
print(Captian_America.power)
```

**In Object Oriented Programming , we are trying to structure the code in a way such that we can match up with the real-world sceanrio.**

```python
# Define a class named Cartoon
# class name first initial is always capital.
# A class is a blueprint used to create objects
class Cartoon:

   # Constructor method (__init__) automatically runs when a new object is created
   # It initializes the attributes of the object
   def __init__(self,name,show,catchphrase):

    # self.name stores the cartoon character's name
    self.name = name

    # self.show stores the cartoon show the character belongs to
    self.show = show

    # self.catchphrase stores the famous dialogue of the character
    self.catchphrase = catchphrase


   # Method to introduce the cartoon character
   # This method prints the character's name
   def introduction(self):
    print(f"Hello My name is {self.name}")


   # Method for the character to say their catchphrase
   def speak(self):
    print(f"{self.name} says {self.catchphrase}")


# Creating an object (instance) of the Cartoon class
# This object represents Mickey Mouse
Mickey_Mouse = Cartoon("Mickey Mouse","Cartoon Network","Oh Boy!")

# Accessing and printing the attribute 'show' of the Mickey_Mouse object
print(Mickey_Mouse.show)

# Calling the introduction method for Mickey Mouse
Mickey_Mouse.introduction()


# Creating another object of the Cartoon class
# This object represents Spongebob
Spongebob = Cartoon("Spongebob","Cartoon Network","Bye Bye!")

# Calling the introduction method for Spongebob
Spongebob.introduction()

# Calling the speak method so Spongebob says his catchphrase
Spongebob.speak()
```

**Object Oriented Programming Terminologies**

**Attributes --> Attributes stands for the characteristicts that a specific object is going to have.**

**Methods --> Functionality that a specific object is going to have.**

**Self --> It refers to the current object of the class.**

**For example : For Avenger class , as soon as we created Iron Man as the object , then self keyword became Iron Man at that point and it helped out to provide the specific attributes to our Iron Man object.**

**But as soon as we created Captian America as the object then self reffered to Captian America to associate the attributes to it.**

**Constructor --> A constructor is being reffered by __init__ .It gets automatically called as soon as we are going to create an object and it is responsible for associating the attributes to your current object.**
