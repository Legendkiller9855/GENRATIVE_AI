```python
# Define a class named Avenger
class Avenger:

  # Constructor method (__init__) runs automatically when an object is created
  # It initializes the attributes of the object
  def __init__(self, name, power, weapon, nationality="American"):

    # self.name stores the name of the Avenger
    self.name = name

    # self.power stores the special power of the Avenger
    self.power = power

    # self.weapon stores the weapon used by the Avenger
    self.weapon = weapon

    # self.nationality stores the nationality of the Avenger
    # "American" is the default value if no nationality is provided
    self.nationality = nationality


  # Method to introduce the Avenger
  # This method prints the details of the Avenger
  def introduction(self):

    # Print the Avenger's name
    print(f"Hello my name is : {self.name}")

    # Print the Avenger's weapon
    print(f"Hello my weapon is : {self.weapon}")

    # Print the Avenger's power
    print(f"Hello my power is : {self.power}")

    # Print the Avenger's nationality
    print(f"Hello my nationality is : {self.nationality}")


# Creating an object of the Avenger class
# Iron_man is an instance (object) of the Avenger class
Iron_man = Avenger("Ironman", "Genius Intellect", "Repulsor Beams")

# Calling the introduction method using the object
Iron_man.introduction()

print(Iron_man.name)

# Key Concepts in This Code

# Class → Blueprint for creating objects

# Constructor (__init__) → Initializes object attributes

# self → Refers to the current object

# Attributes → Variables inside a class (name, power, etc.)

# Method → Function inside a class (introduction)

# Object → Instance of a class (Iron_man)


Captian_America = Avenger("Captian America","Super Strength","Shield","Portugese")
Captian_America.introduction()
print(Captian_America.power)
```

```python
# Define a class named Cartoon
# class name first initial is always capital.
# A class is a blueprint used to create objects
class Cartoon:

   # Constructor method (__init__) automatically runs when a new object is created
   # It initializes the attributes of the object
   def __init__(self,name,show,catchphrase):

    # self.name stores the cartoon character's name
    self.name = name

    # self.show stores the cartoon show the character belongs to
    self.show = show

    # self.catchphrase stores the famous dialogue of the character
    self.catchphrase = catchphrase


   # Method to introduce the cartoon character
   # This method prints the character's name
   def introduction(self):
    print(f"Hello My name is {self.name}")


   # Method for the character to say their catchphrase
   def speak(self):
    print(f"{self.name} says {self.catchphrase}")


# Creating an object (instance) of the Cartoon class
# This object represents Mickey Mouse
Mickey_Mouse = Cartoon("Mickey Mouse","Cartoon Network","Oh Boy!")

# Accessing and printing the attribute 'show' of the Mickey_Mouse object
print(Mickey_Mouse.show)

# Calling the introduction method for Mickey Mouse
Mickey_Mouse.introduction()


# Creating another object of the Cartoon class
# This object represents Spongebob
Spongebob = Cartoon("Spongebob","Cartoon Network","Bye Bye!")

# Calling the introduction method for Spongebob
Spongebob.introduction()

# Calling the speak method so Spongebob says his catchphrase
Spongebob.speak()
```

**Types of Constructors :**


*   Default Constructor --> It doesn't have the parameter inside it and it is only having the self keyword inside it.It assigns a default value to all the objects.

```python
#DEFAULT CONSTRUTOR :
class Dog:
   def __init__(self):
    self.name = "Labrador"
    self.color = "Black"

   def show(self):
    print(f"My name is : {self.name} and my color is :{self.color}")

dog1 = Dog()
dog1.show()

print(dog1.name)

print(dog1.color)
```

**Types of Methods:**

*   Instance Method --> Instance methods operate on object attributes.The instance method is going to use self keyword to access and modify object attributes.

*   Static Method --> Static method do not use self or cls.

They work like normal functions but belong to a class.

They are going to be designed by a abbreviation @staticmethod.

Static method can also be called by the classname also.

*   Class Method --> It is the method of the whole class or the whole blueprint which will remain the same for all of the objects then such kind of method is being reffered as a class method.

In class methods , these methods work with the class level attributes.

They are going to use cls instead of self.

For specific that the current method is a class method , we have to use the abbreviation as @classmethod.



```python
#Instance Method :
class Avenger:
  def __init__(self,name,power):
    self.name = name
    self.power = power

  def show_details(self):
    print(f"My name is {self.name} and I am having power : {self.power}")

Ironman = Avenger("Ironman" , "Genius Mind")
Ironman.show_details()

Thor = Avenger("Thor" , "Hammer")
Thor.show_details()
```

```python
# CLASS METHOD EXAMPLE

# Define a class named Avengers
class Avengers:

  # Class variable (shared by all objects)
  origin = "Avenger"

  # Constructor to initialize object-specific attributes
  def __init__(self,name,power):

    # Instance variable (unique for each object)
    self.name = name

    # Instance variable (unique for each object)
    self.power = power


  # Class method (works with class, not individual object)
  # cls refers to the class itself (Avengers)
  @classmethod
  def origin_info(cls):

    # Access class variable using cls
    print(f"My origin is {cls.origin}")


  # Instance method (works with object)
  # self refers to the current object
  def self_info(self):

    # Access instance variables
    print(f"My name is : {self.name} and my power is : {self.power}")


# Creating an object of Avengers class
Ironman = Avengers("Ironman" , "Genius Intellect")


# Accessing instance variable directly
print(Ironman.power)


# Calling instance method (uses self)
Ironman.self_info()


# Calling class method
# Even though called using object, it works on class variable
Ironman.origin_info()
```

```python
# STATIC METHOD EXAMPLE

# Define a class named Avengers
class Avengers:

  # Static method
  # Does NOT use self (object) or cls (class)
  # Used for general functionality not dependent on object or class data
  @staticmethod
  def mission():
    print("All Avengers must protect the Earth.")


  # Class variable (shared by all objects of the class)
  origin = "Avenger"


  # Constructor method (runs when object is created)
  def __init__(self,name,power):

    # Instance variable (unique for each object)
    self.name = name

    # Instance variable (unique for each object)
    self.power = power


  # Class method
  # Uses cls to access class-level data
  @classmethod
  def origin_info(cls):
    print(f"The origin for the avenger is : {cls.origin}")


  # Instance method
  # Uses self to access object-specific data
  def introduction(self):
    print(f"My name is {self.name} and power is that i have : {self.power}")



# Creating an object (instance) of Avengers class
Ironman = Avengers("Ironman", "Genius Intellect")


# Calling instance method (uses object data)
Ironman.introduction()


# Calling static method using object
Ironman.mission()


# Calling static method using class
Avengers.mission()
```

**PILLARS OF OBJECT ORIENTED PROGRAMMING :**


**1st PILLAR: INHERITANCE**

**Inheritance --> Inheritance is a fundamental concept in object oriented programming because of which the child class is able to inherit the properties or attributes and functionalities of the parent class.**



**Just like a child is going to inherit the properties from parent and parent is going to inherit the properties from the grandparent and definietly in this heirarchy the grandchild can have some features different from the parent and grandparent as well.**


**In Python typically , we get to see 3 types of inheritance :**

*   Single Inheritance
*   Multilevel Inheritance
*   Multiple Inheritance

**Single Inheritance --> The features or the attributes from one parent class will be getting inherited into one child class.**

**Parent Class (Properties --> Attributes and Functionalities)** -----> **Child Class**

**SYNTAX :**

class   child_class_name(parent_class_name):

```python
# SINGLE LEVEL INHERITANCE EXAMPLE

# Parent class (Base class)
class Avengers:

  # Constructor to initialize attributes
  def __init__(self,name,power):

    # Instance variable for name
    self.name = name

    # Instance variable for power
    self.power = power


  # Method to introduce the Avenger
  def introduction(self):
    print(f"My name is : {self.name}")



# Child class (inherits from Avengers)
class Ironman(Avengers):

  # Constructor of child class
  def __init__(self,name,power,weapon):

    # super() calls the parent class constructor
    # This initializes name and power from Avengers class
    super().__init__(name,power)

    # New attribute specific to Ironman class
    self.weapon = weapon


  # Method specific to Ironman class
  def weapon_details(self):
    print(f"I use {self.weapon}")



# Creating an object of Ironman class
Iron = Ironman("Ironman","Genius Intellect","Repulsor Beans")


# Accessing inherited attribute from parent class
print(Iron.name)


# Accessing child class attribute
print(Iron.weapon)


# Calling method defined in child class
Iron.weapon_details()

# Calling method defined in parent class
Iron.introduction()
```

**Multi-Level Inheritance --> The properties or methods and attributes of the grandparent class can be inherited by the parent class and than the properties of the parent class will be inherited by the child class.In a proper hierarchy,the inheritance process is going to work for us.**



**Grandparent(Properties)** --> **Parent(Properties)** --> **Child**

```python
# MULTI-LEVEL INHERITANCE EXAMPLE
# Multi-level inheritance means a class inherits from a parent class,
# and another class inherits from that child class.

# GRANDPARENT CLASS (Base class)
class Grandparent:

  # Constructor initializes attributes of the Grandparent class
  def __init__(self,nationality,baldness,anger_issues):

    # Store nationality of the grandparent
    self.nationality = nationality

    # Store whether grandparent has baldness
    self.baldness = baldness

    # Store whether grandparent has anger issues
    self.anger_issues = anger_issues


  # Method to introduce the grandparent
  def introduction(self):
    print(f"Hello my nationality is : {self.nationality} and about my anger issues if you want to ask about , then the answer is : {self.anger_issues}")



# PARENT CLASS (inherits from Grandparent)
class Parent(Grandparent):

  # Constructor of Parent class
  def __init__(self,nationality,baldness,anger_issues,intelligence):

    # super() calls the constructor of the parent class (Grandparent)
    # This initializes nationality, baldness and anger_issues
    super().__init__(nationality,baldness,anger_issues)

    # New attribute specific to Parent class
    self.intelligence = intelligence


  # Method to display intelligence information
  def intelligence_details(self):
    print(f"About my intelligence if you guys want to know whether if I am intelligent or not , then the answer is : {self.intelligence}")



# CHILD CLASS (inherits from Parent)
class Child(Parent):

  # Constructor of Child class
  def __init__(self,nationality,baldness,anger_issues,intelligence,charm):

    # Calls Parent constructor which also calls Grandparent constructor
    super().__init__(nationality,baldness,anger_issues,intelligence)

    # New attribute specific to Child class
    self.charm = charm


  # Method to display charm details
  def charm_details(self):
    print(f"About my charm if you guys want to know that whether if I am charming or not , then the answer is : {self.charm}")



# Creating an object of Parent class
parent_1 = Parent("Indian","Yes","No","Yes")

# Calling inherited method from Grandparent class
parent_1.introduction()



# Creating an object of Child class
child_1 = Child("Indian","No","Yes","Yes","Yes")

# Calling method from Parent class
child_1.intelligence_details()

# Calling method from Child class
child_1.charm_details()

# Calling inherited method from Grandparent class
child_1.introduction()



# Changing attribute value after object creation
child_1.nationality = "Christian"

# Calling method again to show updated nationality
child_1.introduction()


```

**MULTIPLE INHERITANCE --> In Multiple inheritance , the properties of two parent class or multiple parent class will be getting inherited into the child class.**


**Parent_1(properties)+Parent_2(properties)+Parent_3(properties)+....--> Child**

```python
# MULTIPLE INHERITANCE EXAMPLE
# First Parent Class
class Ironman:

  # Constructor initializes attributes
  def __init__(self,name,power,suit,weapon,nationality = "American"):
    self.name = name
    self.power = power
    self.suit = suit
    self.weapon = weapon
    self.nationality = nationality

  # Method from Ironman class
  def introduction(self):
    print(f"My name is {self.name} and my powers are :{self.power}")


# Second Parent Class
class CaptianAmerica:

  # Constructor initializes attributes
  def __init__(self,name,power,suit,weapon,nationality = "American"):
    self.name = name
    self.power = power
    self.suit = suit
    self.weapon = weapon
    self.nationality = nationality

  # Method from CaptainAmerica class
  def weapon_details(self):
    print(f"I carry the weapon as : {self.weapon}")

# Child Class inheriting from BOTH classes
class Hulk(Ironman, CaptianAmerica):

  # Constructor of child class
  def __init__(self,name,power,suit,weapon,anger_issues,nationality = "American"):

    # super() calls the constructor of the FIRST parent class (Ironman)
    # because of Method Resolution Order (MRO)
    super().__init__(name,power,suit,weapon,nationality)

    # New attribute specific to Hulk class
    self.anger_issues = anger_issues


  # Method specific to Hulk class
  def anger_description(self):
    print(f"About my anger issues if I am having or not then the answer is : {self.anger_issues}")


# Creating object of Hulk class
hulk_guy = Hulk("Hulk","Amazing Power and Strength","Mark 85 Ironman Suit","Repulsor Beans and Shield","Yes","American")

# Calling method from Ironman class (inherited)
hulk_guy.introduction()

# Calling method from CaptainAmerica class (inherited)
hulk_guy.weapon_details()

# Calling method from Hulk class
hulk_guy.anger_description()
```

**2nd PILLAR : POLYMORPHISM**

**Polymorphism --> Polymorphism means many forms in greek.In OOPs , Polymorphism allow objects of different classes as if they are object of the same class and it enables reusability and makes program more flexible.**


**Types of Polymorphism :**

1.   Method Overriding (Runtime Polymorphism)

2.   Method Overloading (Compile Time Polymorphism)

3.   Operator Overloading



**Method Overriding --> A child class provides a specific implementation of a method that has already been defined into the parent class.This allows dynamic selection at runtime.**

```python
# METHOD OVERRIDING + POLYMORPHISM EXAMPLE

# Grandparent class (base class)
class Grandparent:

  # Method speak defined in Grandparent
  def speak(self):
    print("Hello My name is Krishan!")


# Parent class inheriting from Grandparent
class Parent(Grandparent):

  # Overriding the speak() method
  def speak(self):
    print("Hello My name is Junior Krishan!")


# Child class also inheriting from Grandparent
class Child(Grandparent):

  # Overriding the speak() method
  def speak(self):
    print("Hello My name is Junior Krishan Junior!")


# Function that accepts any object and calls its speak() method
def attribute_provider(objects):

  # Calls the speak() method of whatever object is passed
  return objects.speak()


# Creating object of Grandparent
granda = Grandparent()

# Passing Grandparent object → calls Grandparent.speak()
attribute_provider(granda)


# Creating object of Parent
parent = Parent()

# Passing Parent object → calls Parent.speak() (overridden method)
attribute_provider(parent)


# Creating object of Child
child = Child()

# Passing Child object → calls Child.speak() (overridden method)
attribute_provider(child)
```

```python
class Shape:
  def area(self):
    return "This method returns the area of the shape."

class Rectangle(Shape):
  def __init__(self,width,height):
    self.width = width
    self.height = height

  def area(self):
    return self.width*self.height

class Circle(Shape):
  def __init__(self,radius):
    self.radius = radius

  def area(self):
    return 3.14*self.radius*self.radius

shape = Shape()
print(shape.area())

# Corrected instantiation: provide width and height to the constructor
rectangle = Rectangle(6,4)
# Corrected call: area method now uses instance attributes
print(rectangle.area())

circle = Circle(5)
print(circle.area())
```
