```python
#TASK - 1 : Creating Numpy Arrays
import numpy as np
array_1 = np.arange(1,11)

array_2 = np.arange(1,10).reshape(3,3)

array_3 = np.array([10,20,30,40,50])

#Arrays
print(array_1)
print(array_2)
print(array_3)

#Shape of each array
print(array_1.shape)
print(array_2.shape)
print(array_3.shape)

#Data type of each array
print(array_1.dtype)
print(array_2.dtype)
print(array_3.dtype)
```

```python
#TASK - 2 : Important Mathematical Operations
import numpy as np
A = np.array([10,20,30,40])
B = np.array([1,2,3,4])

#Mathematical Operations
print(A + B)
print(A - B)
print(A * B)
print(A / B)
print(A ** 2)

#Optional  :
array_1 = np.add(A,B)
print(array_1)


array_2 = np.subtract(A,B)
print(array_2)
```
