**The default graph type in matplotlib is the line graph**

**If no type of graph is justified , than the type of graph which would be constructed is the line graph.**

**plt.figure() ---> figure() ---> parameter ---> figsize ---> (len,breadth) ---> figure dimension of our graph.**

**plt.figure(figsize = (10,6))**


**plt.title() ---> title() ---> It takes the value as the Title of your plot that what it is representing as what insight we are getting through this graph.**

**SYNTAX : plt.title("Title of Graph" , color = "red")**

**parameter ---> color ---> What color the title is going to have**

**plt.xlabel() ---> Helps to represent that what X axis is representing to us.**


**plt.ylabel() ---> Helps to represent that what Y axis is representing to us.**

**plt.grid() ---> Helps to bring grid on our visualization.**


**plt.show() ---> Helps in representing the chart.**

**plt.xlim() ---> X values ---> [10,1000] ---> [100,1000] ---> [700,800] ---> It puts a limitation that how much the data must be represented on the X axis.**

**plt.ylim() ---> Y values ---> [10,1000] ---> [100,1000] ---> [700,800] ---> It puts a limitation that how much the data must be represented on the Y axis.**

#Cutomization for line in line graph

**PARAMETERS :**


1.   color ---> What is going to be the color of the line ---> for e.g. ---> red , green , blue etc.
2.   marker ---> how point of intersection appears ---> o , ^ , s , * , d , h , p

3.  line style ---> It will be defining the style of the line ---> '--' , '-.'

4.  linewidth ---> It shows the sidth of the line ---> 1 , 1.5 , 2 , 3

5.  markersize ---> It shows the markersize or it can change the size of the marker , default ---> 6

6.  label ---> label that what this specific line is representing but is only helpful while plotting multiple lines in a single visualization




```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

#PANDAS VISUALIZATION :
data = pd.read_csv('/content/mental_health_risk_dataset.csv')
print(data.sample(3))

information = data['gender'].value_counts()

x = information.index
y = information.values

plt.plot(x, y)
plt.title("PANDAS VISUALIZATION")
plt.show()

#MATPLOTLIB VISUALIZATION :
matplotlib_data = data['gender'].value_counts()
plt.figure(figsize = (10,4))
plt.title("Count of Population as per gender" , color = "red")
plt.ylabel("Count of Population")
plt.xlabel("Gender")
plt.grid()
plt.plot(x,y , color = 'purple' , marker = 'o' , linestyle = '--' , linewidth = 2.5 , markersize = 8 , label = 'Gender population')
plt.show()
```
