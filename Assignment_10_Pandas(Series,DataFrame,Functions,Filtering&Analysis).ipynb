```python
#TASK - 1 : PANDAS SERIES BASIC

#importing Pandas as pd
import pandas as pd

#Create a Panda Series from the list.
marks = [78,85,90,66,72]

#Print Series Values
series = pd.Series(marks)
print(series)
print(series.values)

#Print Index
print(series.index)

#Print DataType
print(series.dtype)

#Access First Element
print(series[0])

#Access Last Two Elements
print(series[-2:])
```

```python
#TASK - 2 : MATHEMATICAL OPERATIONS ON SETS

import pandas as pd
marks = [78,85,90,66,72]
series = pd.Series(marks)

#Add 5 grace marks to all students.
new_series = series + 5
print(new_series)

#Subtract 2 marks from all values.
new_subtracted_series = series - 2
print(new_subtracted_series)

#Multiply all marks by 1.05
multiplied_marks = series * 1.05
print(multiplied_marks)

#Divide all marks by 2
divided_series = series/2
print(divided_series)
```

```python
#TASK - 3 : PYTHON FUNCTIONALITIES ON SERIES


import pandas as pd
marks = [78,85,90,66,72]
series = pd.Series(marks)

#Find Maximum Marks
max_marks = series.max()
print(max_marks)

#Find Minmum Marks
min_marks = series.min()
print(min_marks)

#Find Sum of Marks
sum_marks = series.sum()
print(sum_marks)

#Find Mean Marks
mean_marks = series.mean()
print(mean_marks)

#Apply a lambda function to check whether each student has passed(>=70)
condition = series.apply(lambda x: x>=70)
print(condition)

#Count how many students passed
print(condition.sum())
```

```python
#TASK - 5 : IMPORTANT DATAFRAME FUNCTIONS

import pandas as pd
students = {'Name':['Amit','Neha','Rahul','Sneha','Pooja'],
            'Marks':[78,85,90,66,72],
            'Subject':['Math','Math','Science','Science','Math']}
dataframe = pd.DataFrame(students)
print(dataframe)

#Use and print output of:

#info()
print(dataframe.info())

#describe()
print(dataframe.describe())

#head()
print(dataframe.head())

#tail
print(dataframe.tail())

#Sort students by marks in descending order.
ascending_order = dataframe.sort_values('Marks')
descending_order = ascending_order[::-1]
print(descending_order)

#Reset index after sorting
reset_index = descending_order.reset_index(drop=True)
print(reset_index)
```

```python
#TASK - 6 : FILTERING & CONDITIONAL SELECTION

import pandas as pd
students = {'Name':['Amit','Neha','Rahul','Sneha','Pooja'],
            'Marks':[78,85,90,66,72],
            'Subject':['Math','Math','Science','Science','Math']}
dataframe = pd.DataFrame(students)
print(dataframe)

#Students who scored more than 75 marks.
condition_1 = dataframe[dataframe['Marks'] > 75]
print(condition_1)

#Students belonging to subject Math.
condition_2 = dataframe[dataframe['Subject'] == 'Math']
print(condition_2)

#Students who scored more than the average marks.
average_marks = dataframe['Marks'].mean()
condition_3 = dataframe[dataframe['Marks'] > average_marks]
print(condition_3)

#Students who failed (marks < 70).
condition_4 = dataframe[dataframe['Marks'] < 70]
print(condition_4)
```

```python
#TASK - 7 : GROUPING & BASIC ANALYSIS

import pandas as pd
students = {'Name':['Amit','Neha','Rahul','Sneha','Pooja'],
            'Marks':[78,85,90,66,72],
            'Subject':['Math','Math','Science','Science','Math']}
dataframe = pd.DataFrame(students)
print(dataframe)

#Find the average marks
grouped_data = dataframe.groupby('Subject')['Marks']
print(grouped_data)

print(grouped_data.mean())

#Count number of students per subject
print(grouped_data.count())

#Find maximum marks per subject
print(grouped_data.max())
```

```python
#TASK - 8 : PANDAS PLOTTING(SIMPLE GRAPHS)

import pandas as pd
import matplotlib.pyplot as plt # Import matplotlib for showing plots

students = {'Name':['Amit','Neha','Rahul','Sneha','Pooja'],
            'Marks':[78,85,90,66,72],
            'Subject':['Math','Math','Science','Science','Math']}
dataframe = pd.DataFrame(students)
print(dataframe)

#Plot a bar graph of student names vs marks
dataframe.plot(kind='bar', x='Name', y='Marks', title='Marks by Student')
plt.show()

#Plot a line graph of marks
dataframe.plot(kind='line', x='Name', y='Marks', title='Marks Trend by Line Graph')
plt.show()

#Plot a histogram of marks
dataframe.plot(kind='hist', x='Name', y='Marks', title='Marks Trend by Histogram')
plt.show()
```

```python
#TASK - 9 : MINI USE CASE : SALES DATA ANALYSIS

import pandas as pd
import matplotlib.pyplot as plt # Import matplotlib for showing plots

sales = {'Day' : ['Mon','Tue','Wed','Thu','Fri'],
         'Revenue' : [1200,1500,900,2000,1800]
         }
dataframe = pd.DataFrame(sales)
print(dataframe)

#Total Revenue
total_revenue = dataframe['Revenue'].sum()
print(total_revenue)

#Average Daily revenue
average_daily_revenue = dataframe['Revenue'].mean()
print(average_daily_revenue)

#Day with the highest Revenue
day_with_highest_revenue = dataframe.loc[dataframe['Revenue'].idxmax(), 'Day']
print(day_with_highest_revenue)

#Days where revenue > average
revenue_greater_than_average = dataframe[dataframe['Revenue'] > average_daily_revenue]
print(revenue_greater_than_average)

#Plot revenue vs day
plot_graph = dataframe.plot(kind = 'bar' , x = 'Day' , y = 'Revenue' , title = 'Revenue vs Day')
print(plot_graph)
```
