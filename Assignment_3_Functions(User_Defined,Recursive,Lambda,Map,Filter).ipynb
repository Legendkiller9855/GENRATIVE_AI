**TASK - 1 : Basic functions (Price after discount)**

```python
#Write a function apply_discount(price,discount_percent) that:
#Returns the price after discount
#If discount_percent is missing , apply a default discount of 5%
#Test the function with:
#apply_discount(1000,10)
#apply_discount(500)#Uses default discount

# Add a condition inside the function to ensure discount never exceeds 60%
def apply_discount(price, discount_percent=5):
  if discount_percent > 60:
    print(f"Warning: Discount of {discount_percent}% exceeds maximum allowed (60%). Applying 60% discount instead.")
    discount_percent = 60 # Cap the discount at 60%

  discounted_price = price - (discount_percent * price) / 100
  return discounted_price

# Test cases as requested:
print(f"Price after 10% discount on 1000: {apply_discount(1000, 10)}")
print(f"Price after default 5% discount on 500: {apply_discount(500)}")
print(f"Applying 61% discount to 1000 (should cap at 60%): {apply_discount(1000, 61)}")
print(f"Applying 60% discount to 1000: {apply_discount(1000, 60)}")
```
