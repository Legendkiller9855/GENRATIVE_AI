**TASK - 3 : Product Pricing(Dictionaries)**

```python
#Create a dictionary price_dict where keys are product names and values are prices (integers or float) .Include at least 6 entries.
price_dict = {
    "water":23.23,
    "soap":45.2,
    "hair":34,
    "comb":78,
    "brush":98,
    "towel":23
}
print(price_dict)

#Write small code blocks to :
#Add a new product with price to price_dict
#Update the price of an existing product
#Remove a product name (handle the case when the product does not exist )

price_dict["banana"]=12.1
print(price_dict)

price_dict["soap"] = 23.456
print(price_dict)

product_to_remove = "milk"

removed_price = price_dict.pop(product_to_remove, None)

if removed_price is None:
    print("Product not found in dictionary.")
else:
    print(f"{product_to_remove} removed successfully.")

#Print the average price of all products(use only dictionary operations and basic arithmetic)
product_prices = list(price_dict.values())
if product_prices:
    average_price = sum(product_prices) / len(product_prices)
    print(f"Average price of all products: {average_price:.2f}")
else:
    print("No products in the dictionary to calculate average price.")


#EXTRA(Optional : )
#Print the product with both minimum and maximum prices
if product_prices:
    max_price = max(product_prices)
    min_price = min(product_prices)

    max_price_products = [name for name, price in price_dict.items() if price == max_price]
    min_price_products = [name for name, price in price_dict.items() if price == min_price]

    print(f"Product(s) with maximum price ({max_price}): {', '.join(max_price_products)}")
    print(f"Product(s) with minimum price ({min_price}): {', '.join(min_price_products)}")
else:
    print("No products in the dictionary to find min/max prices.")
```

**TASK - 4: Combined Operations**

```python
#Using the products list and price_dict , create a list of tuples named catalog where each tuple is (product_name,price,category)
products = ["comb","brush","toothpaste","soap","towel","water"]
print(products)

price_dict = {
    "comb":23.23,
    "brush":45.2,
    "toothpaste":34,
    "soap":78,
    "towel":98,
    "water":23
}

categories = [
    "Personal Care",
    "Personal Care",
    "Personal Care",
    "Personal Hygiene",
    "Utility",
    "Natural Resource"
]

print(price_dict)

price_dict_keys = price_dict.keys()
print(price_dict_keys)

price_dict_values = price_dict.values()
print(price_dict_values)

catalog = []

for i in range(len(products)):
  product_name = products[i]
  price = price_dict[product_name]
  category = categories[i]


  catalog.append((product_name,price,category))

print(catalog)


#From catalog , create a new dictionary category_to_products that maps each category to a list of product names in that category.

category_to_products = {}

for product, price, category in catalog:
    if category not in category_to_products:
        category_to_products[category] = []

    category_to_products[category].append(product)

print(category_to_products)

#Print all products that belong to the category that has the maximum number of products.

max_category = max(category_to_products, key=lambda k: len(category_to_products[k]))

print("Category with maximum products:", max_category)
print("Products:", category_to_products[max_category])
```
