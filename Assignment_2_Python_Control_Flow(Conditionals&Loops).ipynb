**TASK - 1 : Discount Rules( if / elif / else )**

```python
#Write a program that reads an integer order_amount from the user using input()
order_amount_str = input("Enter a value :")
print(order_amount_str)

#Apply the following discount rules and print the final amount:
#order_amount >= 2000 --> 15% discount
#1500 <= order_amount < 2000 --> 10% discount
#1000<= order_amount < 1500 --> 7% discount
#otherwise --> 0% discount

#Ensure you convert input to a number and handle the case when the input is non-numeric by displaying an error message and exiting the program.


try:
  order_amount = int(order_amount_str)
  if order_amount >= 2000:
   final_amount = order_amount * 0.85 # 15% discount
  elif 1500 <= order_amount < 2000:
   final_amount = order_amount * 0.90 # 10% discount
  elif 1000 <= order_amount < 1500:
   final_amount = order_amount * 0.93 # 7% discount
  else:
   final_amount = order_amount # 0% discount
  print(f"Your final order amount after the applicable discount is : {final_amount:.2f}")
except ValueError:
  print("The given input is not-numeric and is not suitable for calculations.")

#Optional(Add tax(5%) after discount and print subtotal ,tax,and final total (still within conditions and basic arithmetic)

print(final_amount)
taxes = final_amount * 0.05
subtotal = final_amount
amount_after_taxes = final_amount + taxes
print(f"Subtotal is {subtotal} , tax is {taxes} and the final total is {amount_after_taxes}")
```

**TASK - 3 : User Menu(while loop + break/continue)**


```python
orders = []

while True:
    print("\nMenu:")
    print("1. Add order amount")
    print("2. Show all orders and totals after applying discount")
    print("q. Quit")

    choice = input("Enter your choice: ")

    # Add order
    if choice == "1":
        amount = input("Enter order amount: ")

        if not amount.replace('.', '', 1).isdigit():
            print("Invalid input. Please enter a valid number.")
            continue

        orders.append(float(amount))
        print("Order added successfully.")

    # Show orders and totals
    elif choice == "2":
        if not orders:
            print("No orders available.")
            continue

        total = 0
        print("\nOrders with discount applied:")

        for order in orders:
            if order >= 1000:
                discounted_price = order * 0.9   # 10% discount
            else:
                discounted_price = order

            print(f"Original: {order} | After Discount: {discounted_price}")
            total += discounted_price

        print(f"Total after discount: {total}")

    # Quit
    elif choice.lower() == "q":
        print("Exiting program...")
        break

    # Invalid input
    else:
        print("Invalid choice. Try again.")
        continue
```

**TASK - 4 : Loop Control with conditions(break & continue)**

```python
#Given a list of daily sales: daily = [200,150,0,400,50,-1,300] , iterate throughout it with a for loop and
#If a day's value is -1 ,treat it as corrupted data and break the loop (stop processing)
#If a day's value is 0 ,treat it as a day with no sales and continue(skip adding to revenue.)
#For valid positive sales,add to total_sales and print the running total.

daily = [200,150,0,400,50,-1,300]
total_sales = 0
for i in daily:
  if i == -1:
    print("The given data is a corrupted data.")
    break
  if i == 0:
   print("Today's sale is zero sales day.")

  if i > 0 :
   total_sales += i
print(f"Total sales is {total_sales}")

#Print the final total after the loop completes(or stops due to corrupted data.)

daily = [200,150,0,400,50,-1,300]
total_sales = 0
for i in daily:
  if i == -1:
    print("The given data is a corrupted data.")
    continue
  if i == 0:
   print("Today's sale is zero sales day.")

  if i > 0 :
   total_sales += i
print(f"Total sales is {total_sales}")

```
