**PANDAS Series --> Pandas series is a column of data in a table like in excel or google sheets with labels called as indexes attached to each value.**

```python
# Import required libraries
import numpy as np
import pandas as pd

# Creating a Pandas Series
# Data values → prices of items
# index → names of grocery items
# name → label/title of the series

Data = pd.Series(
    [10, 20, 30, 40, 50, 60],   # values
    index=['Bread', 'Butter', 'Milk', 'Tobacco', 'Sugar', 'Coffee'],  # labels
    name='Price of Grocery Items'   # name of the series
)

# Print the Series
print(Data)
```

**Series is a 1-D Array that can hold any type of data like integers , floats ,string(object) , boolean , mixed data type.**

**Data type --> Integers --> Data in a Series(Column) --> Quantity of items in the Stock.**

**Data Type --> Float --> Data in a Series(Column) --> Price or Product Ratings**

**Data Type --> Strings --> Data in a Series(Column) --> Product Name**

**Data Type --> Boolean --> Data in a Series(Column) --> Is product in stock or not**

**Data Type --> Dates/Times --> Data in a Series(Column) --> Expiry date or Sales Date**

**Mixed Type --> Milk , 40 , True**





**Creating a series in multiple Ways :**

*   List
*   Numpy Array
*   Dictionary
*   Custom Index
*   Scalar Value

```python
# Creating Series in Pandas

import numpy as np
import pandas as pd


# ---------------- SERIES FROM LIST ----------------

# List of countries → default index (0,1,2,...)
country = ['India','Germany','USA','China']
print(pd.Series(country))


# List of runs → default index
batsman_runs = [10,20,356,36,76,89]
print(pd.Series(batsman_runs))


# Custom index + datatype
subject_names = ['biology','chemistry','physics','math']
subject_marks = [78,87,89,98]

# Assigning custom labels + setting datatype to int32
print(pd.Series(data=subject_marks, index=subject_names, dtype='int32'))


# ---------------- SERIES FROM NUMPY ARRAY ----------------

# Creating numpy array from 10 to 20
values = np.arange(10,21)

# Converting numpy array to pandas Series
print(pd.Series(values))


# ---------------- SERIES FROM DICTIONARY ----------------

# Dictionary stores data in key-value pairs
# Keys automatically become index, values become data

data_items = {'Maths':78,'Bio':98 , 'Phy':67,'Chem':98}
print(pd.Series(data_items))


# ---------------- CUSTOM INDEX ----------------

# Creating Series with custom labels
values = [1,2,3,4,5]
print(pd.Series(data=values, index=['a','b','c','d','e']))


# ---------------- SCALAR VALUE ----------------

# Single value (4) is assigned to all index positions
print(pd.Series(data=4, index=['a','b','c','d','e']))
```
