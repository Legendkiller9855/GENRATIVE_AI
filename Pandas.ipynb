```python
# Creating Series in Pandas

import numpy as np
import pandas as pd


# ---------------- SERIES FROM LIST ----------------

# List of countries → default index (0,1,2,...)
country = ['India','Germany','USA','China']
print(pd.Series(country))


# List of runs → default index
batsman_runs = [10,20,356,36,76,89]
print(pd.Series(batsman_runs))


# Custom index + datatype
subject_names = ['biology','chemistry','physics','math']
subject_marks = [78,87,89,98]

# Assigning custom labels + setting datatype to int32
print(pd.Series(data=subject_marks, index=subject_names, dtype='int32'))


# ---------------- SERIES FROM NUMPY ARRAY ----------------

# Creating numpy array from 10 to 20
values = np.arange(10,21)

# Converting numpy array to pandas Series
print(pd.Series(values))


# ---------------- SERIES FROM DICTIONARY ----------------

# Dictionary stores data in key-value pairs
# Keys automatically become index, values become data

data_items = {'Maths':78,'Bio':98 , 'Phy':67,'Chem':98}
print(pd.Series(data_items))


# ---------------- CUSTOM INDEX ----------------

# Creating Series with custom labels
values = [1,2,3,4,5]
print(pd.Series(data=values, index=['a','b','c','d','e']))


# ---------------- SCALAR VALUE ----------------

# Single value (4) is assigned to all index positions
print(pd.Series(data=4, index=['a','b','c','d','e']))
```

**Mathematical Functions in Pandas**



*   sum() --> sum of all values present in the series
*   product() --> product of all values in a series.
*   mean() --> Mean provides us with the average value of the data column.

**For E.g. --> Company Data --> Salary --> Average value for that column salary = 6,6,8,9,11**
**mean = (6+6+8+9+11)/5 = 8LPA**

*   median() --> Median also provides the idea about the average of the data , for coming up with median , we need to sort the values in ascending manner and then the middle value after sorting the data --> median , data --> salary --> 1 crore , 10 crore(outlier) --> avg -->  (6+6+8+10+100)/5 = 26LPA**

*   mode() --> Mode provides us with the idea that which value is getting repeated maximum number of times.

*   std() --> If I need to get the idea that how the values inside the series are distributed from the average or mean value then we can come up with the standard deviation.

**SALARY --> 7,8,6,6,8,7,7.5 --> low standard deviation(All values are close to 7)**

**SALARY --> 7,18,26,16,8,17,27.5 --> high standard deviation(All values are far away from 7)**

*  variance() --> Variance also provides the insight about the distribution of the data from the mean value and variance is the square of the standard deviation.

**standard_deviation = |mean -  v_1| + |mean -  v_2| ..../total count of values**

**Variance == std ** 2**

*  min() --> minimum value in the series

*  max() --> maximum value in the series

*  describe() --> describe function and provides the idea about the summary of the important mathematical calculations for the data.

**SERIES METHODS**

1.   head() --> Provides the top preview of data as the top 5 rows of the dataframe or series.It is quite helpful with providing the preview of the dataset and if we want to preview top 10 rows then we can pass 10 as a argument or for top 7 rows we can pass 7 as an argument.

```python
import pandas as pd
import numpy as np


data_path = '/40000_yt_videos.csv'
df = pd.read_csv(data_path) # Read the CSV into a DataFrame

# print(df.head(10)) #For top 10 rows

# print(df.tail(10)) #For bottom 10 rows

# print(df.sample(5))

dm = df["views"]

print(dm)

print(type(dm))

print(dm.sort_values)

## parameter --> ascending = False (sorting --> descending)

print(dm.sort_values(ascending = False))

##METHOD CHAINING --> We will use multiple methods together to get the required values in a way that first method's output will be second method's input and so on.

print(dm.sort_values(ascending = False).head(1))

print(dm.sort_values(ascending = False).head(1).values[0])

#In order to make a change in our data permanent --> inplace

average_data = dm.copy()
print(average_data)
print(average_data.sort_values(inplace = True))

print(dm.sort_index())
```
